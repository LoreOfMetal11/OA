### Часть 4.1 **Pandas** и **Matplotlib**

```python
rain_by_city = data.dropna(subset=["Место", "Осадки"])

rain_by_city = (
    rain_by_city.groupby("Место", as_index=False)["Осадки"]
    .mean()
)

rain_by_city = (
    rain_by_city.sort_values("Осадки", ascending=False)
    .head(15)
)

rain_by_city = rain_by_city.sort_values("Осадки")

plt.figure(figsize=(11, 7))

plt.hlines(
    y=rain_by_city["Место"],
    xmin=0,
    xmax=rain_by_city["Осадки"],
    color="skyblue",
    linewidth=3,
    alpha=0.8
)

plt.scatter(
    x=rain_by_city["Осадки"],
    y=rain_by_city["Место"],
    color="royalblue",
    s=140,
    zorder=3
)

for city, rain in zip(rain_by_city["Место"], rain_by_city["Осадки"]):
    plt.text(
        x=rain + 0.05,
        y=city,
        s=f"{rain:.2f} мм",
        va="center",
        fontsize=10
    )

plt.title("Топ-15 локаций Австралии по среднему количеству осадков")
plt.xlabel("Среднее количество осадков за день, мм")
plt.ylabel("Локация")
plt.grid(axis="x", linestyle="--", alpha=0.35)

plt.tight_layout()
plt.show()
```

График показывает, что осадки в Австралии распределены неравномерно: Cairns, Darwin и CoffsHarbour заметно выбиваются вперёд с показателями 5.74, 5.09 и 5.06 мм в день, тогда как остальные двенадцать локаций образуют плотную группу в диапазоне примерно от 3.0 до 3.8 мм, где значения убывают плавно и различия между соседними позициями минимальны. Из этого следует, что климат определяется прежде всего географией и широтой, а не статусом или размером города: крупные туристические и деловые центры вроде Sydney, Brisbane, Gold Coast и Newcastle оказались в середине или внизу списка, а лидируют тропические и субтропические регионы - северный Квинсленд, Северная территория и влажное побережье Нового Южного Уэльса. Практически это означает, что для сухой погоды стоит выбирать Сидней, Брисбен или Ньюкасл, а для водозависимых отраслей - север и северо-восток; при этом разброс между основной массой локаций невелик (около 25% между 4-м и 15-м местом), а сильный отрыв только трёх лидеров говорит о «тяжёлом хвосте» распределения. Главное ограничение вывода в том, что это среднесуточные значения: они скрывают сезонность, ведь в Darwin почти все осадки приходятся на сезон дождей, а в Cairns - на лето, поэтому для реальных решений нужен разрез по месяцам, иначе среднее может вводить в заблуждение.

В отличие от влажности, где выбросы были только слева, здесь они с двух сторон - распределение симметричнее. Из-за этого среднее значение будет близко к медиане (примерно 22–24 °C), но всё равно чуть смещено. Медиана в 23 °C - надёжный ориентир: половина дней теплее, половина холоднее. Практически это значит, что типичный день около 23 °C, но возможны как заморозки, так и почти 50-градусная жара.

```python
numeric_data = data.select_dtypes(include=np.number)

sns.heatmap(numeric_data.corr(), annot=True, cmap='RdYlGn', linewidths=0.2)

fig=plt.gcf()

fig.set_size_inches(12,12)
plt.show()
```

Минимальная температура имеет очень сильную положительную связь с температурой в 9 часов (r = 0,90) и сильную связь с температурой в 15 часов (r = 0,71). Максимальная температура особенно тесно связана с температурой в 15 часов (r = 0,98), а также сильно коррелирует с температурой в 9 часов (r = 0,89). Это означает, что при повышении утренней, минимальной или максимальной температуры обычно повышается и температура в другие периоды суток.

Испарение также положительно связано с температурой: его коэффициент корреляции с максимальной температурой составляет r = 0,59, а с температурой в 15 часов - r = 0,57. Следовательно, в более тёплые дни испарение, как правило, усиливается.
Влажность, напротив, имеет отрицательную связь с температурой. Например, влажность в 9 часов отрицательно коррелирует с максимальной температурой (r = -0,50), температурой в 9 часов (r = -0,47) и температурой в 15 часов (r = -0,50). Ещё сильнее эта закономерность выражена для влажности в 15 часов: связь с температурой в 15 часов составляет r = -0,56, с максимальной температурой - r = -0,51, а с количеством солнечных часов - r = -0,63. Это говорит о том, что тёплая и солнечная погода обычно сопровождается снижением относительной влажности воздуха.

При этом влажность в 9 и 15 часов положительно связана между собой (r = 0,67). Следовательно, если утром влажность высокая, то велика вероятность, что днём она также останется относительно высокой.

Количество солнечных часов имеет заметную отрицательную зависимость от облачности: с облачностью в 9 часов коэффициент корреляции равен r = -0,68, а с облачностью в 15 часов - r = -0,70. Это означает, что при увеличении облачности количество солнечного света в течение дня уменьшается.

Кроме того, облачность положительно связана с влажностью: коэффициенты корреляции составляют r = 0,45 для влажности в 9 часов и r = 0,52 для влажности в 15 часов. Таким образом, пасмурная погода чаще сопровождается более высокой влажностью воздуха. Между облачностью в 9 и 15 часов также наблюдается положительная связь (r = 0,60), то есть облачная погода утром нередко сохраняется и днём.

Скорость ветра и скорость порывов ветра имеют положительную связь. Наиболее заметна зависимость между скоростью ветра и скоростью порывов в 15 часов (r = 0,69), а также между скоростью ветра и скоростью порывов (r = 0,61). Это означает, что усиление постоянного ветра обычно сопровождается более сильными порывами.

При этом связь ветра с температурой, осадками и облачностью в основном слабая. Поэтому ветер в данной выборке нельзя считать главным фактором, определяющим эти показатели.
