## *Задание* 6. График функции.

Нужно создать и построить график функции `y = cox (x)`, где `x` в интервале от 0 до 10.

```python
x = np.linspace(0, 10, 512)
y = np.cos(x)
plt.plot(x, y)
plt.title("График")
plt.grid(True)
plt.show()
```

## *Задание* 8. Работа с DataFrame.

Нужно создать из готового словаря создать диаграмму и вывести ее.

```python
data = {'Город': ['Москва', 'Казань', 'Томск'], 'Население': [12500000, 1200000, 500000]}
df = pd.DataFrame(data)
df.plot(kind='bar', x='Город', y='Население')
plt.title("Население городов")
plt.show()
```
